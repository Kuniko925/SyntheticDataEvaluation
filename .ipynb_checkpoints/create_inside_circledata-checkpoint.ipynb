```python
import numpy as np
import pandas as pd
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
import data
```

```python
root = "../../dataset/CIFAKE/"
model_names = ['ResNet', 'MobileNet', 'ViT16']
train_settings = ['real', 'fake']
test_settings = ['real', 'fake']

label_to_class = data.get_id_to_class()
target_names = list(label_to_class.values())
```

```python
def save_classification_report(model_name, train_setting, test_setting):
    filepath = f'{root}/results/{model_name}_{train_setting}_{test_setting}.csv'
    save_filepath = f'{root}/results/{model_name}_{train_setting}_{test_setting}_report.csv'
    
    df = pd.read_csv(filepath)
    y_true = df["label"]
    y_pred = df["preds"]
    
    report = classification_report(y_true, y_pred, target_names=target_names, output_dict=True)
    df_report = pd.DataFrame(report)
    df_report = df_report.T.reset_index()
    df_report.rename(columns={'index':'class'}, inplace=True)
    df_report.to_csv(save_filepath, index=False)
    print(f'{model_name}_{train_setting}_{test_setting}')
    print(df_report)
```

```python
for m in model_names:
    for tr in train_settings:
        for te in test_settings:
            save_classification_report(m, tr, te)
```

# Fake - Real Analysis

```python
train_setting = 'fake'
test_setting = 'real'
embeds = ['CLIP', 'DINOv2']
reducers = ['UMAP', 'TSNE']
```

```python
def save_classification_report_by_circle(df, model_name, embed, reducer, inside_circle=True):
    save_filepath = f'{root}/results/{model_name}_circle_{inside_circle}_{embed}_{reducer}.csv'
    df_inside = df[df["inside circle"] == inside_circle]
    y_true = df_inside["label"]
    y_pred = df_inside["preds"]
    
    report = classification_report(y_true, y_pred, target_names=target_names, output_dict=True)
    df_report = pd.DataFrame(report)
    df_report = df_report.T.reset_index()
    df_report.rename(columns={'index':'class'}, inplace=True)
    df_report.to_csv(save_filepath, index=False)
    report
```

```python
def save_distance_dataframe(root, df, model_name, embed, reducer):

    clip_filepath = f'{root}distance/dis_{embed}_{reducer}.csv'
    df_clip = pd.read_csv(clip_filepath)
    df = pd.merge(df, df_clip, on=['label'], how='left', suffixes=('', '_red'))
    
    embed_filepath = f'{root}distance/embed_{embed}_{reducer}.csv'
    df_embed = pd.read_csv(embed_filepath)
    df = pd.merge(df, df_embed, on='filepath', how='left', suffixes=('', '_embed'))
    
    df["distance from real"] = ((df["embeddings x"] - df["r centroid x"])**2 + 
                                    (df["embeddings y"] - df["r centroid y"])**2) ** 0.5
    df["inside circle"] = df["distance from real"] <= df["r radius"]
    
    save_filepath = f'{root}results/{model_name}_{embed}_{reducer}.csv'
    df.to_csv(save_filepath, index=False)

    for inside_circle in [True, False]:
        save_classification_report_by_circle(df, model_name, embed, reducer, inside_circle)
```

```python
for model_name in model_names:
    filepath = f'{root}/results/{model_name}_{train_setting}_{test_setting}.csv'
    df = pd.read_csv(filepath)
    df["correct"] = df["label"] == df["preds"]
    
    for embed in embeds:
        for reducer in reducers:
            save_distance_dataframe(root, df, model_name, embed, reducer)
```

# All data

```python
def plot_correlation_between_distance():
    model_to_color = {
        'ResNet': 'red',
        'MobileNet': 'blue',
        'ViT16': 'green'
    }

    embed_to_style = {
        'CLIP': 'solid',
        'DINOv2': 'dashed'
    }

    reducer_to_marker = {
        'UMAP': 'o',
        'TSNE': '^'}
    
    plt.figure(figsize=(10, 4))

    for model_name in model_names:
        for embed in embeds:
            for reducer in reducers:
                filepath = f'{root}results/{model_name}_{embed}_{reducer}.csv'
                df = pd.read_csv(filepath)
                df["distance bin"] = pd.qcut(df["distance from real"], q=15, duplicates='drop')
                acc_by_distance = df.groupby("distance bin", observed=True)["correct"].mean().reset_index()
                acc_by_distance["bin center"] = acc_by_distance["distance bin"].apply(lambda x: x.mid)

                label = f'{model_name}-{embed}-{reducer}'
                color = model_to_color.get(model_name, 'black') 
                linestyle = embed_to_style.get(embed, 'solid')
                marker = reducer_to_marker.get(reducer, '')
                
                plt.plot(
                    acc_by_distance["bin center"],
                    acc_by_distance["correct"],
                    label=label,
                    linewidth=1,
                    color=color,
                    linestyle=linestyle,
                    marker=marker,
                    markersize=3
                )

    plt.xlabel("Distance from Real Image Centroid")
    plt.ylabel("Accuracy")
    xticks = np.arange(0, acc_by_distance["bin center"].max() + 0.2, 0.2)
    plt.xticks(xticks, rotation=45, fontsize=8)
    plt.yticks(fontsize=8)

    plt.legend(bbox_to_anchor=(1.005, 1), loc='upper left', borderaxespad=0., fontsize=8)
    plt.tight_layout()
    plt.show()
```

```python
plot_correlation_between_distance()
```

```python
%load_ext autoreload
%autoreload 2


import numpy as np
import pandas as pd
import computing
from data import CIFAKEDataSet
```

```python
root = "../../dataset/CIFAKE/cifake1/"
cifake = CIFAKEDataSet(root)

d_type = "train"
save_filepath = f'{root}train.csv'
df_train = cifake.create_dataset(d_type, save_filepath)

entropies = []
for _, row in df_train.iterrows():
    entropy = computing.compute_rgb_joint_entropy(row['filepath'])
    entropies.append(entropy)
df_train['entropy'] = entropies
df_train.to_csv(save_filepath, index=False)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import classification_report
```

```python
label_to_class={0: 'airplane',
             1: 'automobile',
             2: 'bird',
             3: 'cat',
             4: 'deer',
             5: 'dog',
             6: 'frog',
             7: 'horse',
             8: 'ship',
             9: 'truck'}
```

```python
target_names = list(label_to_class.values())
```

```python
def create_report(root, model_name, train_setting, test_setting):

    filepath = f'{root}/results/{model_name}_{train_setting}_{test_setting}.csv'
    save_filepath = f'{root}/results/{model_name}_{train_setting}_{test_setting}_report.csv'
    df = pd.read_csv(filepath)
    labels = df['label']
    preds = df['preds']
    report = classification_report(labels, preds, target_names=target_names, output_dict=True)
    report = pd.DataFrame(report).transpose()
    report.to_csv(save_filepath)
```

```python
root = "../../dataset/CIFAKE/"
model_name = 'ResNet'
train_setting = 'real'
test_setting = 'real'
create_report(root, model_name, train_setting, test_setting)
```
